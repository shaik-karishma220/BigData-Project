# Exploratory Data Analysis (EDA)

This notebook analyzes the Gold layer tables and generates business insights for the Konami Retail project.

```python
#load the gold tables
spark.sql("USE CATALOG konami_catalog")
spark.sql("USE SCHEMA retail_schema")

regional_df = spark.table("gold_regional_market_performance")
product_df = spark.table("gold_product_performance")
shipping_df = spark.table("gold_shipping_analysis")
```

```python
# Preview the Gold Tables

display(regional_df)

display(product_df)

display(shipping_df)
```

```python
# View Table Schemas

regional_df.printSchema()

product_df.printSchema()

shipping_df.printSchema()
```

```python
# Regional Performance
display(regional_df)
```

## Business Insight – Actual Revenue by Region

This chart compares the actual revenue generated by each region.

- It helps identify the highest-performing regions.
- It highlights regions that require additional business attention.
- The analysis supports management in making regional sales and investment decisions.

```python
# Revenue Target Achievement by Region

display(
    regional_df.select(
        "region_name",
        "revenue_target_percentage"
    )
)
```

## Business Insight – Revenue Target Achievement

This chart compares the percentage of the revenue target achieved by each region.

- Regions with higher percentages performed closer to or exceeded their planned revenue targets.
- Regions with lower percentages indicate opportunities to improve sales and marketing performance.
- This analysis helps management evaluate regional performance against business objectives.

# Product Performance Analysis

This section analyzes product sales, revenue, and pricing across different product categories.

```python
display(product_df)
```

## Business Insight – Units Sold by Product Category

This chart compares the number of units sold across different product categories.

- Categories with higher unit sales indicate stronger customer demand.
- Categories with lower sales may require promotional strategies or pricing improvements.
- This analysis helps the business understand customer purchasing trends.

```python
 #Revenue by Product Category
 display(
    product_df.select(
        "category",
        "total_revenue"
    )
)
```

## Business Insight – Revenue by Product Category

This chart compares the total revenue generated by each product category.

- Categories with higher revenue contribute significantly to overall business growth.
- Comparing revenue with units sold helps identify high-value product categories.
- The insights support pricing, inventory, and product planning decisions.

# Shipping Performance Analysis

This section evaluates shipping efficiency and delivery performance across different countries.

```python
display(
    shipping_df.select(
        "shipping_country",
        "average_shipping_days"
    )
)
```

## Business Insight – Average Shipping Days

This chart compares the average shipping time across different countries.

- Lower shipping days indicate faster deliveries.
- Higher shipping times may require improvements in logistics and supply chain operations.

```python
#Delivered Orders
display(
    shipping_df.select(
        "shipping_country",
        "delivered_orders"
    )
)
```

## Business Insight – Delivered Orders

This chart shows the number of successfully delivered orders in each country.

- Countries with more delivered orders indicate stronger operational performance.
- This helps evaluate distribution efficiency.

```python
#Cancelled Orders
display(
    shipping_df.select(
        "shipping_country",
        "cancelled_orders"
    )
)
```

## Business Insight – Cancelled Orders

This chart compares cancelled orders across countries.

- High cancellation rates may indicate delivery or customer service issues.
- The business can investigate these regions to reduce future cancellations.

```python
#Delivary Success Route
display(
    shipping_df.select(
        "shipping_country",
        "delivery_success_rate"
    )
)
```

## Business Insight – Delivery Success Rate

This chart compares the delivery success rate across different countries.

- Higher success rates indicate efficient shipping operations.
- Lower success rates highlight regions where logistics improvements may be needed.
